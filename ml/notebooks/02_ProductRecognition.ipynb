```python
from ultralytics import YOLO
from pathlib import Path

model = YOLO("../saved_models/product_detector.pt")

test_folder = Path("../datasets/products/test/images")

test_images = list(test_folder.glob("*.jpg"))[:2]

for image in test_images:
    results = model.predict(
        source=str(image),
        conf=0.25,
        save=True
    )

    print(image.name)
```

```python
results = model.predict(
    "../datasets/products/test/images/rgb_19_png.rf.08d174759d11c5fb9fdf3d55e4e83027.jpg",
    conf=0.25
)

for box in results[0].boxes:
    cls = int(box.cls[0])
    conf = float(box.conf[0])

    print(
        model.names[cls],
        round(conf, 3)
    )
```
